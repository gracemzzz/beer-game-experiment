## 3. Run the verification — this is the only paid cell

The first four requests check actual ordering decisions under control/trained × minimal/low reasoning with structured output. Then four full games run. All roles use the same weekly snapshot before orders are committed. Incorrect observation/rules replies get up to two retries (three attempts total). The first valid reply is accepted; all attempts and costs remain saved. After exhausted correctness retries, the failed four-role group is saved and a fresh group starts in the same condition. Each target group allows up to three replacement groups. Refused/incomplete responses and permanent API errors stop; temporary rate/network/server errors get bounded transport retries. No invented orders are used. The progress bar includes access checks and rules checks.

If stopped, the next cell still reports what happened. Do not keep rerunning a failed request or spend the rest of your $3 without reviewing it. A spending stop can resume successful cached requests if you deliberately raise only BUDGET_USD; a saved failed response requires review.

Independent groups and same-week roles run concurrently, with at most 4 requests in flight. The bar shows completed/remaining requests, active requests, and each group's week. On a terminal failure, no new requests are sent; already-sent requests finish and save their usage before stopping. The total spending reservation includes every in-flight request. If interrupted, unknown usage stays reserved. Temporary rate-limit/network/server errors use up to five transport retries, a shared cooldown, and adaptive slower request starts. Confirmed billing/quota errors stop immediately.

Rules-check clarification: `total_holding_cost` explicitly means the charge for all units, not the per-unit rate. Each candidate group must pass its four rules checks before its game begins. The correct answer is not supplied to the model or substituted by code. Incorrect answers are logged with expected/received values and may be retried twice. This new protocol preserves the earlier failed pilot; clarification may help but does not guarantee minimal reasoning is reliable.

**Bounded retry policy:** three attempts maximum per logical request. Incorrect observation/rules replies get up to two retries. Separately, temporary rate/network/server errors get up to five transport retries without prompt changes. Authentication, confirmed billing/quota, refusal and token-limit errors stop. Retries keep the model, reasoning effort, game state and history fixed, with a generic recheck reminder; they do not supply the correct answer or a preferred order. Every attempt counts against $0.75. Resuming the same run reuses saved attempts and cannot reset the retry limit. The progress total includes retries, and the report shows recovery counts by condition.


```python
client = None
results = None
run_dir = study.latest_run(BASE, CONFIG)
try:
    if MODE == "api":
        from openai import AsyncOpenAI
        key = os.environ.get("OPENAI_API_KEY") or getpass.getpass("OpenAI API key (hidden): ")
        if not key.strip():
            raise ValueError("An API key is required in API mode.")
        client = AsyncOpenAI(api_key=key.strip(), max_retries=0, timeout=CONFIG["timeout_s"])
        del key
    results, run_dir = await study.run_study(BASE, CONFIG, client=client)
    print("VERIFICATION COMPLETE. Run the next cell for the results.")
except study.ExperimentStopped as exc:
    print(str(exc))
    print("Run the next cell to review the saved status. Replacement limit, spending guard, or non-retryable error stopped the run. No automatic model upgrade.")
finally:
    if client is not None:
        await client.close()
    print("Results folder:", run_dir)
```
